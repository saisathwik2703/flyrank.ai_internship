**Question:** Which pages should be reviewed first for content refresh?

**Lane:** Refresh / Content Opportunity Scoring

> Scope: the metrics used here are the repository's verified 30,000-row starter benchmark. This notebook does not claim a fresh full-warehouse training run.


## 1. Question

Rank content pages for human editorial review so limited review capacity is spent on pages with the strongest measurable evidence of decline/opportunity. The output is decision support, not an automated claim that a refresh will cause recovery.

## 2. Data

Warehouse release: `flyrank_pseudonymized_warehouse_release_v20260703` (export 2026-07-03). The daily fact table has 78,835,655 rows covering 2025-01-27 to 2026-06-30; dimension tables contain pseudonymized client/content IDs. Raw client names, domains, URLs, queries, titles, and keywords are excluded. The benchmark results below are from the 30,000-row anonymized starter slice.

## 3. Methodology

Starter label: `is_declining_label = trend_direction == 'down'` (a current-window proxy). The transparent baseline combines visibility, freshness risk, position opportunity, and depth gap. Logistic regression, decision tree, and random forest are evaluated against it on the same client-holdout split. IDs/product decisions are excluded and future-window leakage is prohibited.

```python
import pandas as pd
results = pd.DataFrame({
    'method':['Baseline rules','Logistic regression','Decision tree','Random forest'],
    'roc_auc':[0.627,0.700,0.742,0.750],
    'average_precision':[0.468,0.522,0.575,0.618],
    'precision_at_50':[0.240,0.400,0.540,0.740]
})
results
```

## 4. Results (vs baseline)

The verified starter benchmark improves Precision@50 from 0.24 for the rule baseline to 0.74 for the random forest. That corresponds to roughly 12 vs 37 positives among the first 50 review candidates (≈3.1× top-50 precision). ROC AUC improves from 0.627 to 0.750.

```python
import matplotlib.pyplot as plt
ax = results.set_index('method')['precision_at_50'].plot(kind='bar', legend=False)
ax.set_ylabel('Precision@50')
ax.set_xlabel('Method')
ax.set_title('Model vs baseline on the verified starter benchmark')
plt.tight_layout()
plt.show()
```

## 5. Limitations

- The reported metrics are from the starter slice, not a new full-warehouse benchmark.
- The starter label is a current-window proxy rather than a future-window outcome.
- The warehouse is an unbalanced panel with different client tracking start dates.
- Observational ranking cannot prove that a content edit causes recovery.
- Consolidation, seasonality, SERP changes, and low-volume noise can mimic decline.

## 6. Ranked recommendations

1. Review visible pages with sustained decline and meaningful demand first.
2. Protect page-one performers showing decay risk after seasonality/consolidation checks.
3. Investigate low CTR only with sufficient impressions and position context.
4. Use staleness/thinness as reason codes, not standalone proof.
5. Keep human review and monitor drift, coverage, precision, base rate, and feature distributions.
